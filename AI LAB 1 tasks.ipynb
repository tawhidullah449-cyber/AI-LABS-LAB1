```python

def sundaes():
    flavors = ["vanilla", "chocolate", "strawberry", "pistacchio"]
    sauces = ["caramel", "butterscotch", "chocolate"]

    count = 0

    for flavor in flavors:
        for sauce in sauces:
            print(flavor + " ice cream sundae with " + sauce + " sauce")
            count += 1

    return count


total = sundaes()
print("Total combinations:", total)
```

```python
def count_pattern(pattern, lst):
    count=0

    for i in range(len(lst) - len(pattern)+1):
        if lst[i:i + len(pattern)]==pattern:
           count+=1
    return count
    
l1=("a","b","c","e","b","a","b","f")
print(count_pattern(("a","b"),l1))
```

```python
class solution:
    def tow_sum(self,numbers,target):
        for i in range(len(numbers)):
            for j in range(i+1,len(numbers)):
                if numbers[i]+numbers[j]==target:
                    return i,j

numbers=[10,20,10,40,50,60,70]
target=50
obj=solution()
print("outpu", obj.tow_sum(numbers,target))
```

```python
class solution():
    def three_sum(self,numbers):
        result=[]
        for i in range(len(numbers)):
            for j in range(i+1,len(numbers)):
                for k in range(j+1,len(numbers)):
                    if numbers[i]+numbers[j]+numbers[k]==0:
                       result.append([numbers[i], numbers[j], numbers[k]])
        return result

numbers=[-25,-10,-7,-3,2,4,8,10]
obj=solution()
print("output: ", obj.three_sum(numbers))
```

```python

n = int(input("Enter the size of square matrices: "))

# a. Create lists to store matrix elements
matrix1 = []
matrix2 = []
result = []

# b. Read elements of the first matrix
print("Enter elements of the first matrix:")
for i in range(n):
    row = []
    for j in range(n):
        value = int(input(f"Enter element [{i}][{j}]: "))
        row.append(value)
    matrix1.append(row)

# Read elements of the second matrix
print("Enter elements of the second matrix:")
for i in range(n):
    row = []
    for j in range(n):
        value = int(input(f"Enter element [{i}][{j}]: "))
        row.append(value)
    matrix2.append(row)

# c and d. Add corresponding elements and store in third matrix
for i in range(n):
    row = []
    for j in range(n):
        row.append(matrix1[i][j] + matrix2[i][j])
    result.append(row)

# Display the result
print("Sum of the two matrices:")
for row in result:
    print(row)
```

```python

# a. Read the dimensions of the matrices
r1 = int(input("Enter rows of first matrix: "))
c1 = int(input("Enter columns of first matrix: "))

r2 = int(input("Enter rows of second matrix: "))
c2 = int(input("Enter columns of second matrix: "))

if c1 != r2:
    print("Matrix multiplication is not possible!")
else:
    # Create lists to store matrix elements
    matrix1 = []
    matrix2 = []
    result = []

    # b. Read elements of the first matrix
    print("Enter elements of first matrix:")
    for i in range(r1):
        row = []
        for j in range(c1):
            row.append(int(input()))
        matrix1.append(row)

    # Read elements of the second matrix
    print("Enter elements of second matrix:")
    for i in range(r2):
        row = []
        for j in range(c2):
            row.append(int(input()))
        matrix2.append(row)

    # c and d. Multiply the matrices
    for i in range(r1):
        row = []
        for j in range(c2):
            total = 0
            for k in range(c1):
                total += matrix1[i][k] * matrix2[k][j]
            row.append(total)
        result.append(row)

    # e. Display the result
    print("Result of matrix multiplication:")
    for row in result:
        print(row)
```

```python

def calculator(num1, num2, operation="addition", output_format="float"):
    # Check the operation
    if operation not in ["addition", "subtraction", "multiplication", "division"]:
        raise ValueError("Invalid operation!")

    # Check the output format
    if output_format not in ["integer", "float"]:
        raise ValueError("Invalid output format!")

    # Perform the requested operation
    if operation == "addition":
        result = num1 + num2
    elif operation == "subtraction":
        result = num1 - num2
    elif operation == "multiplication":
        result = num1 * num2
    elif operation == "division":
        if num2 == 0:
            raise ZeroDivisionError("Cannot divide by zero!")
        result = num1 / num2

    # Convert the result to the requested format
    if output_format == "integer":
        return round(result)
    else:
        return float(result)


# Examples
print(calculator(10, 5))
print(calculator(10, 5, "subtraction"))
print(calculator(10, 5, "multiplication", "integer"))
print(calculator(10, 4, "division"))
```

```python

class Numbers:
    MULTIPLIER = 10

    def __init__(self, x, y):
        self.x = x
        self.y = y

    # a. Instance method
    def add(self):
        return self.x + self.y

    # b. Class method
    @classmethod
    def multiply(cls, a):
        return a * cls.MULTIPLIER

    # c. Static method
    @staticmethod
    def subtract(b, c):
        return b - c

    # d. Property
    @property
    def value(self):
        return (self.x, self.y)

    # Setter
    @value.setter
    def value(self, values):
        self.x, self.y = values

    # Deleter
    @value.deleter
    def value(self):
        del self.x
        del self.y


# Create an object
obj = Numbers(20, 5)

print("Addition:", obj.add())
print("Multiplication:", Numbers.multiply(5))
print("Subtraction:", Numbers.subtract(20, 5))
print("Values:", obj.value)

# Change x and y using the setter
obj.value = (30, 10)
print("Updated values:", obj.value)

# Delete x and y using the deleter
del obj.value
```
